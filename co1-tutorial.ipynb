## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">1️⃣ Stage 1 &middot; One Order, One Invoice</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Topics used</div>Variables &bull; data types &bull; operators &bull; f-strings</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ Steps</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Store the order details in variables</li><li>Calculate subtotal, GST and total</li><li>Print a clear invoice</li></ul></div>

```python
# One order, one invoice

customer = "Ravi Kumar"
item = "Notebook"
quantity = 10
unit_price = 45.50
gst_rate = 18                               # percent

subtotal = quantity * unit_price
gst = subtotal * gst_rate / 100
total = subtotal + gst

print(f"Customer : {customer}")
print(f"{quantity} x {item} @ Rs {unit_price} = Rs {subtotal:.2f}")
print(f"GST ({gst_rate}%) : Rs {gst:.2f}")
print(f"Total    : Rs {total:.2f}")
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">2️⃣ Stage 2 &middot; All Orders in One Place</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Topics used</div>Lists &bull; dictionaries &bull; list of dictionaries</div>

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">3️⃣ Stage 3 &middot; A Discount Rule</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Topics used</div>Conditionals &bull; comparison operators</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ The rule</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Subtotal of <b>5,000 or more</b> &rarr; 10% discount</li><li>Subtotal of <b>1,000 or more</b> &rarr; 5% discount</li><li>Otherwise &rarr; no discount</li><li>GST is charged on the amount <b>after</b> discount</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# A discount rule (one order first)

order = orders[3]                                        # Meena Iyer's order
subtotal = order["quantity"] * order["unit_price"]       # 12 x 650 = 7800

if subtotal >= 5000:
    discount_rate = 10
elif subtotal >= 1000:
    discount_rate = 5
else:
    discount_rate = 0

discount = subtotal * discount_rate / 100
taxable = subtotal - discount
gst = taxable * 18 / 100
total = taxable + gst

print(f"Subtotal : Rs {subtotal:.2f}")
print(f"Discount : {discount_rate}% = Rs {discount:.2f}")
print(f"GST (18%): Rs {gst:.2f}")
print(f"Total    : Rs {total:.2f}")
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">4️⃣ Stage 4 &middot; Invoices for Every Order</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Topics used</div>for loop &bull; conditionals &bull; f-string formatting</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ Steps</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Put the Stage 3 logic <b>inside a for loop</b></li><li>Print one invoice per order &ndash; automatically</li><li>Same code, any number of orders</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# Invoices for every order

for order in orders:
    subtotal = order["quantity"] * order["unit_price"]

    if subtotal >= 5000:
        discount_rate = 10
    elif subtotal >= 1000:
        discount_rate = 5
    else:
        discount_rate = 0

    discount = subtotal * discount_rate / 100
    taxable = subtotal - discount
    gst = taxable * 18 / 100
    total = taxable + gst

    print("-" * 40)
    print(f"Customer : {order['customer']}")
    print(f"{order['quantity']} x {order['item']} @ Rs {order['unit_price']}")
    print(f"Subtotal : Rs {subtotal:,.2f}")
    print(f"Discount : {discount_rate}% (Rs {discount:,.2f})")
    print(f"GST      : Rs {gst:,.2f}")
    print(f"TOTAL    : Rs {total:,.2f}")
print("-" * 40)
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">5️⃣ Stage 5 &middot; Clean &amp; Reusable with Functions</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Topics used</div>Functions &bull; return values &bull; constants &bull; clean code</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ Steps</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Move each job into a <b>function</b> with a docstring</li><li><code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">get_discount_rate</code> &rarr; <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">create_invoice</code> &rarr; <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">print_invoice</code></li><li>Functions <b>return</b> values; printing happens separately</li><li>Named constants replace magic numbers</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# Functions

GST_RATE = 18                                  # one place to change the rate

def get_discount_rate(subtotal):
    """Return the discount percentage for a subtotal."""
    if subtotal >= 5000:
        return 10
    elif subtotal >= 1000:
        return 5
    return 0

def create_invoice(order):
    """Return a dictionary with every amount for one order."""
    subtotal = order["quantity"] * order["unit_price"]
    discount_rate = get_discount_rate(subtotal)
    discount = subtotal * discount_rate / 100
    taxable = subtotal - discount
    gst = taxable * GST_RATE / 100
    return {
        "customer": order["customer"],
        "subtotal": subtotal,
        "discount_rate": discount_rate,
        "discount": discount,
        "gst": gst,
        "total": taxable + gst,
    }

def print_invoice(invoice):
    """Print one invoice neatly."""
    print(f"{invoice['customer']:<12} subtotal {invoice['subtotal']:>9,.2f}"
          f"  disc {invoice['discount_rate']:>2}%  gst {invoice['gst']:>8,.2f}"
          f"  TOTAL {invoice['total']:>9,.2f}")

invoices = []
for order in orders:
    invoice = create_invoice(order)
    invoices.append(invoice)
    print_invoice(invoice)
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">6️⃣ Stage 6 &middot; Daily Sales Summary</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Topics used</div>Everything together: loops &bull; accumulators &bull; conditionals &bull; formatting</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ Steps</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Total sales, total discount and total GST (accumulators)</li><li>Find the <b>biggest order</b> (loop + if)</li><li>Print a one-screen summary the owner can read in seconds</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# Daily sales summary

total_sales = 0
total_discount = 0
total_gst = 0
biggest = invoices[0]

for invoice in invoices:
    total_sales += invoice["total"]
    total_discount += invoice["discount"]
    total_gst += invoice["gst"]
    if invoice["total"] > biggest["total"]:
        biggest = invoice

print("=" * 40)
print("DAILY SALES SUMMARY")
print("=" * 40)
print(f"Orders processed : {len(invoices)}")
print(f"Total sales      : Rs {total_sales:>12,.2f}")
print(f"Discounts given  : Rs {total_discount:>12,.2f}")
print(f"GST collected    : Rs {total_gst:>12,.2f}")
print(f"Biggest order    : {biggest['customer']} (Rs {biggest['total']:,.2f})")
print("=" * 40)
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">🏭 Case: Multi-File Commission &amp; Mess-Queue Tallies</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 The problem</div>Two repetitive chores, the kind a back office does by hand every month: <b>(1)</b> pay sales reps commission across several monthly sales &ldquo;files&rdquo;, and <b>(2)</b> tally a hostel mess's weekly meal orders and cost. We model each &ldquo;file&rdquo; as a <b>list of dictionaries</b>, then automate both with a <b>function</b> and loops, rounding to paise.</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ What we will do</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Combine several &ldquo;files&rdquo; and total commission per rep with <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">.get()</code></li><li>Apply a tiered commission (7% at/over Rs 20,000, else 4%) inside a function</li><li>Tally the mess queue: total meals, total cost, average cost per meal, busiest day</li><li>Find the top rep with <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">max(..., key=...)</code></li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# multi-file commission & mess-queue tallies

# Two "files" arrive as separate lists of records (in real life, CSVs you read in later)
north_rep_sales = [
    {"rep": "Asha",  "month": "Jan", "sales": 18500},
    {"rep": "Meena", "month": "Feb", "sales": 24300},
    {"rep": "Raj",   "month": "Mar", "sales": 9800},
]
south_rep_sales = [
    {"rep": "Meena", "month": "Jan", "sales": 31200},
    {"rep": "Kiran", "month": "Feb", "sales": 15600},
]

def commission_for(sales):
    """Return the commission, rounded to paise. 7% at/over 20,000, else 4%."""
    rate = 0.07 if sales >= 20000 else 0.04
    return round(sales * rate, 2)

# Combine every "file" and tally commission per rep
all_sales = north_rep_sales + south_rep_sales
rep_totals = {}
for record in all_sales:
    rep = record["rep"]
    rep_totals[rep] = rep_totals.get(rep, 0) + commission_for(record["sales"])

for rep, amount in sorted(rep_totals.items(), key=lambda kv: kv[1], reverse=True):
    print(f"{rep:<6} commission Rs {amount:>9,.2f}")
print("Total commission paid: Rs", round(sum(rep_totals.values()), 2))
top_rep = max(rep_totals, key=rep_totals.get)
print("Top rep:", top_rep, "Rs", rep_totals[top_rep])

# ---- Mess queue: weekly meal orders and their cost ----
mess_queue = [
    {"day": "Mon", "meals_served": 18, "cost": 950},
    {"day": "Tue", "meals_served": 22, "cost": 1100},
    {"day": "Wed", "meals_served": 16, "cost": 820},
    {"day": "Thu", "meals_served": 25, "cost": 1180},
    {"day": "Fri", "meals_served": 20, "cost": 980},
]
total_meals = 0
total_cost = 0
for day in mess_queue:
    total_meals += day["meals_served"]
    total_cost += day["cost"]
print("\nMess this week:", total_meals, "meals, Rs", total_cost)
print("Average cost per meal: Rs", round(total_cost / total_meals, 2))
busiest_day = max(mess_queue, key=lambda d: d["meals_served"])
print("Busiest day:", busiest_day["day"], "-", busiest_day["meals_served"], "meals")
```
