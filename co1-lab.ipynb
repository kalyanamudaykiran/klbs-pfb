## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">📦 What is a Variable?</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Definition</div>A variable is a <b>name</b> that points to a <b>value</b> stored in the computer's memory. Think of it as a <b>sticky-note label</b> on a box that holds something.</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;text-align:center;margin:14px 0"><span style="font-family:Consolas,Menlo,monospace;background:#FEF3C7;color:#78350F;padding:6px 14px;border-radius:8px;font-size:30px;font-weight:600">unit_price</span> &nbsp;➜&nbsp; <span style="font-family:Consolas,Menlo,monospace;background:#DBEAFE;color:#1E3A8A;padding:6px 14px;border-radius:8px;font-size:30px;font-weight:600">45.50</span></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Key ideas</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Create a variable by <b>assignment</b>: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">name = value</code></li><li>No declaration needed &ndash; Python decides the type from the value (<i>dynamic typing</i>)</li><li>A variable can be <b>re-assigned</b> to a new value at any time</li><li>Read <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">=</code> as <b>&ldquo;gets&rdquo;</b>, not &ldquo;equals&rdquo;: the right side is worked out first, then stored under the name on the left</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# ---------- LECTURE 1 · Slide 3: Variables ----------

# Example 1: Creating variables (assignment uses a single =)
product_name = "Notebook"      # text
unit_price = 45.50             # decimal number
quantity = 12                  # whole number
print(product_name, unit_price, quantity)

# Example 2: Using variables in a calculation, then re-assigning
total_cost = unit_price * quantity
print("Total cost:", total_cost)

quantity = 20                  # the label now points to a NEW value
print("Updated quantity:", quantity)
print("total_cost is still:", total_cost)   # NOT updated automatically!

# Example 3: Multiple assignment and swapping values
shop_a, shop_b = 5, 10         # two variables in one line
shop_a, shop_b = shop_b, shop_a
print(shop_a, shop_b)          # 10 5
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">🏷️ Naming Variables: Rules, Dos &amp; Don'ts</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#FFFBEB;border-left:8px solid #D97706;color:#78350F;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">⚠️ Exceptions (errors) you will meet</div><ul style="margin:4px 0 0 0;padding-left:22px"><li><b>SyntaxError</b> &ndash; illegal name such as <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">class = 5</code></li><li><b>NameError</b> &ndash; using a name that was never created (usually a typo or wrong capital letter)</li></ul></div>

```python
# ---------- LECTURE 1 · Slide 4: Naming rules ----------

# Example 1: Good names - readable and meaningful
customer_name = "Ravi Kumar"
total_price = 2500
is_paid = True
print(customer_name, total_price, is_paid)

# Example 2: Case-sensitivity - these are TWO different variables
price = 100
Price = 999
print(price, Price)            # 100 999

# Example 3: Python's reserved keywords can never be used as names
import keyword
print(keyword.kwlist)

# ❌ Uncomment ONE line at a time to see the error, then comment it again
# 1st_item = "pen"             # SyntaxError: starts with a digit
# my price = 50                # SyntaxError: space in the name
# class = "A"                  # SyntaxError: 'class' is a keyword
# print(prce)                  # NameError: typo - 'prce' was never created
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">🧾 A Neatly Formatted Bill</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Goal</div>Print a tidy bill with aligned columns using f-string formatting.</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ What we will do</div><ul style="margin:4px 0 0 0;padding-left:22px"><li><code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">:.2f</code> &rarr; 2 decimals, <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">:,</code> &rarr; thousands separator</li><li><code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">:&lt;12</code> left-align, <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">:&gt;10</code> right-align in a given width</li><li>Subtotal, GST and total at the bottom</li></ul></div>

```python
# ---------- PRACTICAL 1 · Exercise 4: A Neatly Formatted Bill ----------

# Three items: name, price, quantity
item1_name, item1_price, item1_qty = "Notebook", 45.50, 10
item2_name, item2_price, item2_qty = "Pen", 10, 25
item3_name, item3_price, item3_qty = "Marker", 32.75, 4

# Line amounts and totals
line1 = item1_price * item1_qty
line2 = item2_price * item2_qty
line3 = item3_price * item3_qty
subtotal = line1 + line2 + line3
gst = subtotal * 18 / 100
total = subtotal + gst

# Format specs:  <12 = left in 12 spaces,  >10 = right in 10 spaces,  .2f = 2 decimals
print("=" * 37)
print(f"{'ITEM':<12}{'QTY':>5}{'PRICE':>10}{'AMOUNT':>10}")
print("-" * 37)
print(f"{item1_name:<12}{item1_qty:>5}{item1_price:>10.2f}{line1:>10.2f}")
print(f"{item2_name:<12}{item2_qty:>5}{item2_price:>10.2f}{line2:>10.2f}")
print(f"{item3_name:<12}{item3_qty:>5}{item3_price:>10.2f}{line3:>10.2f}")
print("-" * 37)
print(f"{'Subtotal':<27}{subtotal:>10,.2f}")
print(f"{'GST 18%':<27}{gst:>10,.2f}")
print(f"{'TOTAL':<27}{total:>10,.2f}")
print("=" * 37)
```

```python
# ---------- LECTURE 2 · Slide 4: List methods ----------

# Example 1: Adding and removing items
stock = ["Pen", "Book"]
stock.append("Bag")             # add at the END
stock.insert(1, "Lamp")         # add at position 1
print(stock)                    # ['Pen', 'Lamp', 'Book', 'Bag']
stock.remove("Book")            # remove by VALUE
last = stock.pop()              # remove AND return the last item
print(stock, "| popped:", last)

# Example 2: Numbers - summary functions and sorting
sales = [1200, 450, 3100, 780]
print(sum(sales), min(sales), max(sales), len(sales))
sales.sort()                              # sorts IN PLACE (changes the list)
print(sales)
print(sorted(sales, reverse=True))        # NEW sorted list, original untouched

# Example 3: Alias vs copy
a = [1, 2, 3]
b_alias = a                     # NOT a copy - both names point to ONE list
b_copy = a.copy()               # a real, independent copy
a.append(99)
print(b_alias)                  # [1, 2, 3, 99]  <- changed too!
print(b_copy)                   # [1, 2, 3]

# ❌ Uncomment ONE line at a time to see the behaviour
# stock.remove("Rocket")        # ValueError: not in list
# result = sales.sort()         # sort() returns nothing...
# print(result)                 # ...so this prints None
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">📖 Dictionaries: Key &rarr; Value</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Definition</div>A <b>dictionary</b> stores <b>key : value</b> pairs inside curly braces. You look things up by <b>key</b> &ndash; not by position.</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;text-align:center;margin:14px 0"><span style="font-family:Consolas,Menlo,monospace;background:#FEF3C7;color:#78350F;padding:6px 14px;border-radius:8px;font-size:30px;font-weight:600">"name"</span> &nbsp;➜&nbsp; <span style="font-family:Consolas,Menlo,monospace;background:#DBEAFE;color:#1E3A8A;padding:6px 14px;border-radius:8px;font-size:30px;font-weight:600">"Ravi Kumar"</span></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Key ideas</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Create: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">{&quot;name&quot;: &quot;Ravi&quot;, &quot;orders&quot;: 3}</code></li><li>Read: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">d[&quot;name&quot;]</code> &nbsp; Safe read: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">d.get(&quot;name&quot;, default)</code></li><li>Add <b>or</b> change: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">d[&quot;city&quot;] = &quot;Chennai&quot;</code></li><li>Keys must be <b>unique</b> and unchangeable (text, numbers, tuples)</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# ---------- LECTURE 2 · Slide 5: Dictionaries ----------

# Example 1: Creating a dictionary and reading values by key
customer = {"name": "Ravi Kumar", "city": "Chennai", "orders": 3}
print(customer["name"])         # Ravi Kumar
print(customer.get("city"))     # Chennai

# Example 2: Adding a new key and updating an existing one
customer["phone"] = "90000-00000"   # new key  -> added
customer["orders"] = 4              # old key  -> value replaced
print(customer)

# Example 3: Safe lookups and key checks
print(customer.get("email", "not provided"))   # default instead of an error
print("city" in customer)                      # True  (checks KEYS)
print(len(customer))                           # number of pairs
```

```python
# ---------- LECTURE 2 · Slide 6: Dictionary methods ----------

# Example 1: Looking at keys, values and pairs
prices = {"Pen": 10, "Notebook": 45, "Bag": 499}
print(prices.keys())
print(prices.values())
print(prices.items())

# Example 2: update() and pop()
prices.update({"Lamp": 650, "Pen": 12})   # adds Lamp, changes Pen
print(prices)
removed = prices.pop("Bag")               # remove the key, get its value back
print("Removed price:", removed, "| Left:", prices)

# Example 3: Duplicate keys - the last one wins (no error!)
dup = {"a": 1, "a": 2}
print(dup)                                # {'a': 2}

# ❌ Uncomment ONE line at a time to see the error, then comment it again
# print(prices["Rocket"])                 # KeyError: 'Rocket'
# bad = {["x"]: 1}                        # TypeError: unhashable type: 'list'
```

```python
# ---------- PRACTICAL 2 · Exercise 2: Price List with a Dictionary ----------

price_list = {"Rice": 62, "Dal": 140, "Oil": 155, "Sugar": 44}
print("Rice costs", price_list["Rice"])

price_list["Oil"] = 160                  # price change
price_list["Tea"] = 220                  # new product
del price_list["Sugar"]                  # remove a product
print(len(price_list), "products:", list(price_list.keys()))

# get() avoids a KeyError for products we do not stock
print(price_list.get("Salt", "Not in price list"))

# Price a cart: look every product up in the dictionary
cart = ["Rice", "Dal", "Oil"]
total = 0
for product in cart:                     # for loop preview - Lecture 3 explains it
    total += price_list[product]         # look up the price by key
print("Bill:", total)                    # 62 + 140 + 160 = 362
```

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Goal</div>Work with a small product inventory the way real business data is stored.</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ What we will do</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Read inside records: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">inventory[1][&quot;name&quot;]</code></li><li>Update a quantity and add a new record</li><li>Calculate the stock value of each product and in total</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# PRACTICAL 2 · Exercise 3: Inventory (list of dictionaries)

inventory = [
    {"sku": "P101", "name": "Pen",      "qty": 120, "price": 10},
    {"sku": "N202", "name": "Notebook", "qty": 40,  "price": 45},
    {"sku": "B303", "name": "Bag",      "qty": 8,   "price": 499},
]

# Read: list index first, then dictionary key
print(inventory[1]["name"], "->", inventory[1]["qty"], "in stock")

# Update: restock the bags
inventory[2]["qty"] += 20

# Add: a new product record
inventory.append({"sku": "L404", "name": "Lamp", "qty": 15, "price": 650})

# Value of each product (for loop preview)
for product in inventory:
    value = product["qty"] * product["price"]
    print(f"{product['name']:<10} qty={product['qty']:>4}   value={value:>7}")

# Total stock value - the accumulator pattern
total_value = 0
for product in inventory:
    total_value += product["qty"] * product["price"]
print("Total stock value: Rs", total_value)
```

<pre style="font-family:Consolas,Menlo,monospace;background:#0F172A;color:#E2E8F0;padding:14px 18px;border-radius:8px;font-size:23px;line-height:1.5;margin:10px 0;overflow-x:auto">if condition_1:

```python
# ---------- LECTURE 3 · Slide 3: Conditionals ----------

# Example 1: if / else - one decision
stock = 15
if stock < 20:
    print("Reorder needed")
else:
    print("Stock OK")

# Example 2: if / elif / else - the FIRST True branch wins
marks = 72
if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
else:
    grade = "C"
print("Grade:", grade)

# Example 3: combining conditions with and / or
order_total = 6000
is_member = True
if order_total >= 5000 and is_member:
    print("10% member discount")
elif order_total >= 5000 or is_member:
    print("5% discount")
else:
    print("No discount")
```

    # this block runs once for each item</pre>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Key ideas</div><ul style="margin:4px 0 0 0;padding-left:22px"><li><code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">range(1, 4)</code> gives 1, 2, 3 &ndash; the stop number is <b>not</b> included</li><li><b>Accumulator pattern</b>: start <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">total = 0</code> before the loop, then <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">total += x</code> inside it</li><li>Loop over a dictionary with <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">for key, value in d.items():</code></li></ul></div>
<div style="display:flex;gap:14px;flex-wrap:wrap"><div style="flex:1;min-width:360px"><div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ Do</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Name the loop variable as one item: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">for product in products</code></li></ul></div></div><div style="flex:1;min-width:360px"><div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#FEF2F2;border-left:8px solid #DC2626;color:#7F1D1D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">❌ Don't</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Reset <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">total = 0</code> <b>inside</b> the loop</li><li>Add or remove list items while looping over it</li></ul></div></div></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#FFFBEB;border-left:8px solid #D97706;color:#78350F;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">⚠️ Exceptions (errors) you will meet</div><ul style="margin:4px 0 0 0;padding-left:22px"><li><b>TypeError</b> &ndash; <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">for x in 5</code>: a number is not iterable</li><li><b>NameError</b> &ndash; using <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">total</code> before creating it</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# ---------- LECTURE 3 · Slide 5: for loops ----------

# Example 1: Loop over a list
products = ["Pen", "Notebook", "Bag"]
for product in products:
    print("Packing:", product)

# Example 2: range() - counting; the stop number is NOT included
for day in range(1, 4):              # 1, 2, 3
    print("Day", day)
print(list(range(0, 10, 2)))         # step of 2 -> [0, 2, 4, 6, 8]

# Example 3: The accumulator pattern - running totals
sales = [1200, 450, 3100, 780]
total = 0                            # start BEFORE the loop
for amount in sales:
    total += amount                  # add each amount to the running total
print("Total sales:", total)

# Bonus: looping over a dictionary
prices = {"Pen": 10, "Bag": 499}
for item, price in prices.items():
    print(item, "costs", price)

# ❌ Uncomment to see the error, then comment it again
# for x in 5:                        # TypeError: 'int' object is not iterable
#     print(x)
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">⏳ while Loops, break &amp; continue</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Definition</div>A <b>while loop</b> repeats <b>as long as</b> its condition stays True. Use it when you do not know in advance how many times to repeat.</div>
<pre style="font-family:Consolas,Menlo,monospace;background:#0F172A;color:#E2E8F0;padding:14px 18px;border-radius:8px;font-size:23px;line-height:1.5;margin:10px 0;overflow-x:auto">while condition:
    # do work
    # change something so the condition can become False</pre>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Key ideas</div><ul style="margin:4px 0 0 0;padding-left:22px"><li><code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">break</code> &ndash; leave the loop immediately</li><li><code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">continue</code> &ndash; skip to the next round</li><li>Use <b>for</b> when you have a collection; use <b>while</b> when you have a condition</li></ul></div>
<div style="display:flex;gap:14px;flex-wrap:wrap"><div style="flex:1;min-width:360px"><div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ Do</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Change a variable inside the loop so it can end</li></ul></div></div><div style="flex:1;min-width:360px"><div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#FEF2F2;border-left:8px solid #DC2626;color:#7F1D1D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">❌ Don't</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Write <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">while True:</code> without a <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">break</code></li><li>Forget the update line &ndash; that creates an <b>infinite loop</b></li></ul></div></div></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#FFFBEB;border-left:8px solid #D97706;color:#78350F;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">⚠️ Stuck in an infinite loop?</div>Press the <b>&#9632; Stop</b> button on the cell (or <b>Runtime &rarr; Interrupt execution</b>).</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# ---------- LECTURE 3 · Slide 6: while, break, continue ----------

# Example 1: while - repeat until the condition becomes False
stock = 5
while stock > 0:
    print("Selling... stock left:", stock)
    stock -= 1                       # without this line -> infinite loop!
print("Sold out")

# Example 2: break - stop the loop early
prices = [120, 450, 99, 1500, 30]
for price in prices:
    if price > 1000:
        print("Found a high-value item:", price)
        break                        # no need to look further

# Example 3: continue - skip this round only
for price in prices:
    if price < 100:
        continue                     # skip cheap items
    print("Counting", price)
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">🧰 Functions: Write Once, Use Many Times</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Definition</div>A <b>function</b> is a named block of code. You give it <b>inputs (parameters)</b> and it can <b>return</b> a result.</div>
<pre style="font-family:Consolas,Menlo,monospace;background:#0F172A;color:#E2E8F0;padding:14px 18px;border-radius:8px;font-size:23px;line-height:1.5;margin:10px 0;overflow-x:auto">def function_name(parameter1, parameter2=default):
    &quot;&quot;&quot;One-line description.&quot;&quot;&quot;
    # work happens here
    return result</pre>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Key ideas</div><ul style="margin:4px 0 0 0;padding-left:22px"><li><b>Define</b> once with <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">def</code>, <b>call</b> as many times as you like</li><li>Parameters can have <b>default values</b>: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">rate=18</code></li><li><code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">return</code> sends a value back so you can keep using it</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# ---------- LECTURE 3 · Slide 7: Functions ----------

# Example 1: A function with parameters, a default value and a return
def calculate_gst(amount, rate=18):
    """Return the GST for an amount (default rate is 18%)."""
    return amount * rate / 100

print(calculate_gst(2500))           # uses the default rate -> 450.0
print(calculate_gst(2500, 5))        # rate given -> 125.0

# Example 2: Functions can call other functions
def invoice_total(amount, rate=18):
    """Return the amount plus GST."""
    return amount + calculate_gst(amount, rate)

print(invoice_total(2500))           # 2950.0

# Example 3: A function that only DOES something (no return)
def greet_customer(name):
    print(f"Welcome, {name}!")

greet_customer("Ravi")
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">✨ Function Pitfalls &amp; Clean Code</span>

<div style="display:flex;gap:14px;flex-wrap:wrap"><div style="flex:1;min-width:360px"><div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ Do</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Give each function <b>one job</b></li><li>Use verb names: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">calculate_gst</code></li><li><b>Return</b> results, then print outside</li><li>Add a short docstring</li></ul></div></div><div style="flex:1;min-width:360px"><div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#FEF2F2;border-left:8px solid #DC2626;color:#7F1D1D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">❌ Don't</div><ul style="margin:4px 0 0 0;padding-left:22px"><li><b>Print</b> inside when you needed a value back (you get <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">None</code>)</li><li>Expect a variable created inside a function to exist outside</li><li>Repeat magic numbers such as <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">0.18</code> everywhere</li></ul></div></div></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Clean-code checklist</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Meaningful <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">snake_case</code> names</li><li>Constants in CAPITALS: <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">GST_RATE = 18</code></li><li>Small functions, consistent 4-space indentation</li><li>Comments explain <b>why</b>, not what</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#FFFBEB;border-left:8px solid #D97706;color:#78350F;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">⚠️ Exceptions (errors) you will meet</div><ul style="margin:4px 0 0 0;padding-left:22px"><li><b>TypeError</b> &ndash; missing required argument when you forget an input</li><li><b>NameError</b> &ndash; using a function's inner variable outside it</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# ---------- LECTURE 3 · Slide 8: Function pitfalls and clean code ----------

# Example 1: print vs return
def gst_print(amount):
    print(amount * 0.18)             # shows a value but gives nothing back

def gst_return(amount):
    return amount * 0.18             # gives the value back - reusable!

result_a = gst_print(1000)           # prints 180.0
result_b = gst_return(1000)
print(result_a, result_b)            # None 180.0
print(result_b + 100)                # we can keep calculating with it

# Example 2: Scope - variables created inside a function stay inside
def make_bill(amount):
    bill = amount + 50
    return bill

print(make_bill(200))                # 250

# Example 3: Clean code - named constant, clear names, docstring
GST_RATE = 18                        # one place to change the rate

def price_with_gst(price):
    """Return the price including GST."""
    return price + price * GST_RATE / 100

print(price_with_gst(1000))          # 1180.0

# ❌ Uncomment ONE line at a time to see the error, then comment it again
# print(bill)                        # NameError: 'bill' exists only inside make_bill
# print(price_with_gst())            # TypeError: missing 1 required argument
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">🔀 Discount Decision Maker</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ What we will do</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>A tiered <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">if / elif / else</code> ladder (most specific first)</li><li>A second, separate <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">if</code> for the member bonus</li><li>A formatted result with f-strings</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# ---------- PRACTICAL 3 · Exercise 1: Discount Decision Maker ----------

order_total = 7500
is_member = True

# Tiered discount - the FIRST true branch wins, so start with the biggest
if order_total >= 10000:
    discount_rate = 15
elif order_total >= 5000:
    discount_rate = 10
elif order_total >= 1000:
    discount_rate = 5
else:
    discount_rate = 0

# Members get 2% extra - a separate if, because it is an independent rule
if is_member:
    discount_rate += 2

discount = order_total * discount_rate / 100
print(f"Discount : {discount_rate}% = Rs {discount:.2f}")
print(f"To pay   : Rs {order_total - discount:.2f}")
```

```python
# ---------- PRACTICAL 3 · Exercise 2: Loop Workout ----------

weekly_sales = [12500, 9800, 15200, 11000, 18700, 21000, 8900]

# Part A: total and average (accumulator pattern)
total = 0
for amount in weekly_sales:
    total += amount
average = total / len(weekly_sales)
print("Total:", total, "| Average:", round(average, 2))

# Part B: find the best day (loop + if)
best = weekly_sales[0]
for amount in weekly_sales:
    if amount > best:
        best = amount
print("Best day's sales:", best)

# Part C: while loop - a loan repaid at Rs 12,000 a month
balance = 50000
month = 0
while balance > 0:
    month += 1
    balance -= 12000
    if balance < 0:
        balance = 0                   # never show a negative balance
    print(f"Month {month}: balance Rs {balance}")

# Part D: a multiplication table with range()
for n in range(1, 6):
    print(f"7 x {n} = {7 * n}")
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">🧰 Build Your Own Functions</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Goal</div>Turn the discount and GST logic into small reusable functions and test them.</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ What we will do</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>One job per function, with a docstring</li><li>A function that calls other functions</li><li>A loop that tests the function with several orders</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# ---------- PRACTICAL 3 · Exercise 3: Build Your Own Functions ----------

GST_RATE = 18                            # constant: one place to change

def calculate_gst(amount, rate=GST_RATE):
    """Return the GST on an amount."""
    return amount * rate / 100

def discount_rate_for(total):
    """Return the discount percentage for an order total."""
    if total >= 10000:
        return 15
    elif total >= 5000:
        return 10
    elif total >= 1000:
        return 5
    return 0

def final_bill(total, is_member=False):
    """Return the amount payable after discount and GST."""
    rate = discount_rate_for(total)
    if is_member:
        rate += 2
    discounted = total - total * rate / 100
    return round(discounted + calculate_gst(discounted), 2)

# Test the functions on several orders
for order_total in [800, 2500, 7500, 15000]:
    print(order_total, "->", final_bill(order_total), "| member:", final_bill(order_total, True))
```

```python
# ---------- PRACTICAL 3 · Exercise 4: Clean Code Makeover ----------

# ❌ BEFORE: it works, but what does it do?
x=[250,1200,80,4500]
t=0
for i in x:
  if i>100:
    t=t+i*0.18
print(round(t,2))

# ✅ AFTER: same result, readable by anyone
GST_RATE = 0.18
MIN_TAXABLE_AMOUNT = 100

def total_gst(invoice_amounts):
    """Return the total GST on invoices above the taxable minimum."""
    gst_total = 0
    for amount in invoice_amounts:
        if amount > MIN_TAXABLE_AMOUNT:
            gst_total += amount * GST_RATE
    return gst_total

invoice_amounts = [250, 1200, 80, 4500]
print(round(total_gst(invoice_amounts), 2))
```

## <span style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#3B82F6;font-size:46px">5️⃣ Stage 5 &middot; Clean &amp; Reusable with Functions</span>

<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#EFF6FF;border-left:8px solid #2563EB;color:#1E293B;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">💡 Topics used</div>Functions &bull; return values &bull; constants &bull; clean code</div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;background:#ECFDF5;border-left:8px solid #16A34A;color:#14532D;padding:12px 18px;margin:10px 0;border-radius:6px;font-size:27px;line-height:1.55"><div style="font-weight:700;font-size:30px;margin-bottom:4px">✅ Steps</div><ul style="margin:4px 0 0 0;padding-left:22px"><li>Move each job into a <b>function</b> with a docstring</li><li><code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">get_discount_rate</code> &rarr; <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">create_invoice</code> &rarr; <code style="background:#E2E8F0;color:#0F172A;padding:2px 6px;border-radius:4px;font-size:90%">print_invoice</code></li><li>Functions <b>return</b> values; printing happens separately</li><li>Named constants replace magic numbers</li></ul></div>
<div style="font-family:'Segoe UI', Roboto, Helvetica, Arial, sans-serif;color:#FCA311;font-weight:700;font-size:24px;margin-top:8px">▶ Run the code cell below</div>

```python
# ---------- TUTORIAL 1 · Stage 5: Functions ----------

GST_RATE = 18                                  # one place to change the rate

def get_discount_rate(subtotal):
    """Return the discount percentage for a subtotal."""
    if subtotal >= 5000:
        return 10
    elif subtotal >= 1000:
        return 5
    return 0

def create_invoice(order):
    """Return a dictionary with every amount for one order."""
    subtotal = order["quantity"] * order["unit_price"]
    discount_rate = get_discount_rate(subtotal)
    discount = subtotal * discount_rate / 100
    taxable = subtotal - discount
    gst = taxable * GST_RATE / 100
    return {
        "customer": order["customer"],
        "subtotal": subtotal,
        "discount_rate": discount_rate,
        "discount": discount,
        "gst": gst,
        "total": taxable + gst,
    }

def print_invoice(invoice):
    """Print one invoice neatly."""
    print(f"{invoice['customer']:<12} subtotal {invoice['subtotal']:>9,.2f}"
          f"  disc {invoice['discount_rate']:>2}%  gst {invoice['gst']:>8,.2f}"
          f"  TOTAL {invoice['total']:>9,.2f}")

invoices = []
for order in orders:
    invoice = create_invoice(order)
    invoices.append(invoice)
    print_invoice(invoice)
```

```python
# ---------- TUTORIAL 1 · Stage 6: Daily sales summary ----------

total_sales = 0
total_discount = 0
total_gst = 0
biggest = invoices[0]

for invoice in invoices:
    total_sales += invoice["total"]
    total_discount += invoice["discount"]
    total_gst += invoice["gst"]
    if invoice["total"] > biggest["total"]:
        biggest = invoice

print("=" * 40)
print("DAILY SALES SUMMARY")
print("=" * 40)
print(f"Orders processed : {len(invoices)}")
print(f"Total sales      : Rs {total_sales:>12,.2f}")
print(f"Discounts given  : Rs {total_discount:>12,.2f}")
print(f"GST collected    : Rs {total_gst:>12,.2f}")
print(f"Biggest order    : {biggest['customer']} (Rs {biggest['total']:,.2f})")
print("=" * 40)
```
